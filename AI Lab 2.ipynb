```python
x = int(input("Enter an integer value: "))

if x > 0:
  print("x is positive")
elif x < 0:
  print("x is negative")
elif x == 0:
  print("x is Zero")
else:
  print("Invalid input")
```

```python
x = int(input("Enter an integer value: "))

if x == 0:
  print("x is Zero")
elif x < 0 and (x%2==0):
  print("x is negative and even")
elif x < 0 and (x%2==1):
  print("x is negative and odd")
elif x > 0 and (x%2==0):
  print("x is Positive and even")
elif x > 0 and (x%2==1):
  print("x is Positive and odd")
else:
  print("Invalid input")
```

```python
marks = float(input("Enter your marks: "))
if marks > 100 or marks < 0:
  print("invalid input")
elif marks > 90:
  print("Excellent")
elif marks > 80:
  print("Good")
elif marks > 70:
  print("Average")
else:
  print("Need improvement")
```

```python
name = input("Enter your name: ")

print("Name: ", name)
print("Reverse name: ", name[::-1])
print("Uppercase name: ", name.upper())
print("Lowercase name: ", name.lower())
print("Length of name: ", len(name))
```

```python
name = input("Enter your name: ")

print("First character:", name[0])
print("Last character:", name[-1])
print("Sperated character")
for i in name:
  print(i)
print("Name Length:", len(name))
```

```python
num = 1

print("Printing numbers from 1 to 10")
while num <= 10:
  print(num, " ", end="")
  num+=1

print("\n\nPrinting numbers from 10 to 1")
while num >= 1:
  print(num, " ", end="")
  num-=1

print("\n\nPrint even numbers")
while num <= 20:
  print(num, " ", end="")
  num+=2

```
